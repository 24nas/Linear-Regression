CALCULUS OF THE GRADIENT


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(24)
n= 15
d = 7
X = np.random.rand(n,d)

true_w = np.array([1.5, -2.0, 3.0, 0.5, -1.0, 2.5, -0.5]).reshape(d, 1)
true_b = 4.0

noise = np.random.randn(n, 1) * 0.1
y = (X @ true_w) + true_b + noise

w = np.zeros((d, 1))
b = 0.0

learning_rate = 0.1
num_iterations = 50
```

```python
def y_prediction(w,X,b):
    return X@w +b
```

```python
def lossFunctionOld (X,w,b,y):
    sum = 0
    for i in range (n):
        sum+= (1/n)*np.square(y_prediction[i]- y[i])
    return sum

```

```python
def lossFunction(X,w,b,y):
   return np.mean((y_prediction(w,X,b)-y)**2)
```

```python
def gradientW_L(y, w, X, b):
  y_pred = y_prediction(w, X, b)
  return (2 / n) * X.T @ (y_pred - y)
```

```python
def gradientB_L (y,w,X,b):
    y_pred = y_prediction(w, X, b)
    return (2/n)*np.sum(y_pred - y)
```

```python
def gradient_descent (w,b,X, learning_rate):
    w = w- learning_rate*gradientW_L (y,w,X,b)
    b = b - learning_rate*gradientB_L(y,w,X,b)
    return w,b

```

```python
lossFunction(X,w,b,y)
for i in range (num_iterations):
    w,b=gradient_descent(w,b,X, learning_rate)
    print(lossFunction(X,w,b,y))

```

```python
learning_rates = [0.01, 0.1, 0.5]
num_iterations = 40

plt.figure(figsize=(9, 6))

for lr in learning_rates:
 
  w = np.zeros((d, 1))
  b = 0.0
  loss_history = []

  for i in range(num_iterations):
    loss_history.append(lossFunction(X, w, b, y))
    w, b = gradient_descent(w, b, X, lr)

 
  plt.plot(loss_history, marker='o', label=f'LR = {lr}')

plt.title('Comparaison of different learning rates "(LR)"')
plt.xlabel("Iteration number")
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```
